```python
import functools

def decorator_a(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"A: Before the function {func.__name__}")
        result = func(*args, **kwargs)
        print(f"A: After the function {func.__name__}")
        return result
    return wrapper

def decorator_b(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"B: Before the function {func.__name__}")
        result = func(*args, **kwargs)
        print(f"B: After the function {func.__name__}")
        return result
    return wrapper
```

```python
@decorator_a
@decorator_b
def greetings(name):
    print(f"Hello {name}!")

greetings("Tolgahan")
```
