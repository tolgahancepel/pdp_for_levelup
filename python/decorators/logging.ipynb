```python
import functools
import time
import logging

logging.basicConfig(filename="logs.txt")
logger = logging.getLogger("MyLogger")
logger.setLevel(logging.INFO)
```

```python

def log_decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.perf_counter()

        logger.info(f"START: Function '{func.__name__}' called with args={args}, kwargs={kwargs}")

        try:
            result = func(*args, **kwargs)

            end_time = time.perf_counter()
            duration = end_time - start_time
            logger.info(f"SUCCESS: Function '{func.__name__}' returned {result} | Execution time: {duration:.4f}s")
            # print(f"The duration is: {duration}")
        except Exception as e:
            end_time = time.perf_counter()
            duration = end_time - start_time
            # print(f"ERROR: The duration is: {duration}")
            logger.error(f"ERROR: Function '{func.__name__}' failed with exception: {e} | Execution time: {duration:.4f}s")
            raise
        return result
    return wrapper
```

```python
@log_decorator
def greetings(name):
    print(f"Hello {name}!")
    time.sleep(2)

greetings("Tolgahan")
```
